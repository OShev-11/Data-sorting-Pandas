```python
df = pd.read_csv('movies_stats.csv')
df.head(10)
```

```python
def production_year(row):
    years = range(1950, 2011)

    for year in years:
        if str(year) in str(row):
            return year

    return '1900'

df['year'] = df['title'].apply(production_year)
df.head(10)
```

```python
df['year'] = pd.to_numeric(df['year'], errors='coerce')
```

```python
print(df['year'].dtype)
```

```python
df['year'].unique()
```

```python
df.groupby('year').agg({'rating': 'mean'}).sort_values(by='rating', ascending=False)
```
